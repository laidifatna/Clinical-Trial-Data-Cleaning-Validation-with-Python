### Data Consistency

Let's review the dataset for unexpected or inconsistent values and identify potential extreme observations. Extreme values will be investigated as possible data-entry errors, measurement errors, or valid observations. No clinical measurements will be changed without a documented methodological reason.

```python
# Identify potential outliers using the IQR rule
potential_outliers = []

for i in quantitative_positions:
    values = df.iloc[:, i]

    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outlier_count = ((values < lower_bound) | (values > upper_bound)).sum()

    potential_outliers.append({
        "Column_Position": i,
        "Variable": df.columns[i],
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Potential_Outliers": outlier_count
    })

potential_outliers = pd.DataFrame(potential_outliers)

potential_outliers
```

```python
# Show variables with potential outliers
potential_outliers[
    potential_outliers["Potential_Outliers"] > 0
].sort_values(
    "Potential_Outliers",
    ascending=False
)
```

```python
# Check for actual negative values in quantitative variables
negative_values = []

for i in quantitative_positions:
    values = df.iloc[:, i]
    count = (values < 0).sum()

    if count > 0:
        negative_values.append({
            "Variable": df.columns[i],
            "Negative_Count": count
        })

negative_values = pd.DataFrame(negative_values)

print("Variables containing negative values:")
print(negative_values)
```

```python
# Show the actual negative values
for variable in negative_values["Variable"]:
    values = df.loc[df[variable] < 0, ["Participant_ID", "Treatment", "Measurement_Time", variable]]
    
    print(f"\n{variable}:")
    print(values.to_string(index=False))
```

Negative cytokine values may reflect assay-related effects such as low-level measurements or standard-curve calibration. Previous studies have shown that background subtraction in multiplex immunoassays can produce negative assay values, particularly when measured signals are close to or below the background level. These values were therefore retained rather than treated automatically as data-entry errors.

As regards `ITA_cheek`, negative values are expected for this measurement, while the negative values observed for `Elasticity_F0_cheek` and `Elasticity_F1_forehead` are nearly zero and were therefore also retained.

Conclusion: all negative values will be retained for statistical analysis and will not be removed or modified solely because they are negative.

```python
# Identify potential outliers by treatment group
outliers_by_treatment = []

for i in quantitative_positions:
    values = df.iloc[:, i]

    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outlier_mask = (values < lower_bound) | (values > upper_bound)

    if outlier_mask.sum() > 0:
        for treatment in df["Treatment"].unique():
            count = (
                outlier_mask &
                (df["Treatment"] == treatment)
            ).sum()

            if count > 0:
                outliers_by_treatment.append({
                    "Variable": df.columns[i],
                    "Treatment": treatment,
                    "Potential_Outliers": count
                })

outliers_by_treatment = pd.DataFrame(outliers_by_treatment)

outliers_by_treatment.sort_values(
    "Potential_Outliers",
    ascending=False
)
```

```python
# Calculate the percentage of potential outliers for each variable
outlier_frequency = []

for i in quantitative_positions:
    values = df.iloc[:, i]

    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outlier_mask = (
        (values < lower_bound) |
        (values > upper_bound)
    )

    non_missing = values.notna().sum()
    outlier_count = outlier_mask.sum()

    outlier_percent = (
        outlier_count / non_missing * 100
    )

    outlier_frequency.append({
        "Column_Position": i,
        "Variable": df.columns[i],
        "Non_Missing": non_missing,
        "Potential_Outliers": outlier_count,
        "Outlier_Percent": outlier_percent
    })

outlier_frequency = pd.DataFrame(outlier_frequency)
```

```python
# Divide variables according to the proportion of potential outliers
outlier_frequency["Outlier_Group"] = pd.cut(
    outlier_frequency["Outlier_Percent"],
    bins=[-np.inf, 20, 40, np.inf],
    labels=["<20%", "20%–40%", ">40%"],
    right=False
)

outlier_frequency["Outlier_Group"].value_counts()
```

```python
outlier_frequency["Potential_Outliers"].value_counts().sort_index()
```

```python
# Group variables according to the number of potential outliers
outlier_frequency["Outlier_Group"] = pd.cut(
    outlier_frequency["Potential_Outliers"],
    bins=[-1, 0, 5, 10, 20, np.inf],
    labels=[
        "0",
        "1–5",
        "6–10",
        "11–20",
        ">20"
    ]
)

outlier_frequency["Outlier_Group"].value_counts().sort_index()
```

```python
# Review variables with more than 20 potential outliers
high_outlier_variables = outlier_frequency[
    outlier_frequency["Outlier_Group"] == ">20"
].copy()

# Add outlier counts for each treatment group
for treatment in ["Placebo", "SH60", "SH120"]:
    counts = []

    for _, row in high_outlier_variables.iterrows():
        i = row["Column_Position"]
        values = df.iloc[:, i]

        q1 = values.quantile(0.25)
        q3 = values.quantile(0.75)
        iqr = q3 - q1

        lower_bound = q1 - 1.5 * iqr
        upper_bound = q3 + 1.5 * iqr

        outlier_mask = (
            (values < lower_bound) |
            (values > upper_bound)
        )

        count = (
            outlier_mask &
            (df["Treatment"] == treatment)
        ).sum()

        counts.append(count)

    high_outlier_variables[treatment] = counts

# Display the complete table
print(high_outlier_variables.to_string(index=False))
```
